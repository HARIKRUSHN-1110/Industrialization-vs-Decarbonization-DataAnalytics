## Importing library


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Reading data
We are using the dataset from https://sdi.eea.europa.eu/catalogue/srv/eng/catalog.search#/metadata/21e758c6-a9ac-4a7d-a64a-19d2ba9eecb7

```python
data = pd.read_excel("../data/raw/EEA_Industry_Dataset_EPRTR_Air_Releases.xlsx" , sheet_name="Air_Releases_Sector")
```

```python
print(data.count())
data.head()
```

## Getting some info about the dataset

```python
countries_name = data['countryName'].unique()
print(len(countries_name))
```

## Triming data 
We only need EU countries data and dropping all he unnecessary coloums from the data sets 

```python
countries_EU = ["Germany", "France", "Italy", "Spain" ,"Netherlands" ,"Poland"]
results = data[data['countryName'].isin(countries_EU)].reset_index(drop=True)
results = results.drop(["EPRTR_SectorCode"],axis=1)
```

```python
results.to_excel("../data/processed/Industry_Air_Release.xlsx")
```

## Ploting the datasets 

```python
columns_to_plot = [f"{year}" for year in range(2007, 2023)]
```

```python
fig, axes = plt.subplots(3, 2, figsize=(16, 14))
axes = axes.flatten()

colors = ['#FF6B6B', '#4ECDC4', '#45B7D1','#FF6B6B', '#4ECDC4', '#45B7D1','#FF6B6B', '#4ECDC4', '#45B7D1']
for i, country in enumerate(countries_EU):
    country_data = results[results['countryName'] == country]
    gas_types = country_data['Pollutant'].unique()
    
    years = [int(year) for year in columns_to_plot]
    x = np.arange(len(years))
    width = 0.8 / len(gas_types)
    
    # Plot bars for each gas type
    for j, gas in enumerate(gas_types):
        gas_row = country_data[country_data['Pollutant'] == gas]
        values = gas_row[columns_to_plot].values.flatten().astype(float)
        
        offset = (j - len(gas_types)/2) * width + width/2
        
        axes[i].bar(x + offset, values, width, label=gas, 
                   color=colors[j], alpha=0.8, edgecolor='black')
    
    axes[i].set_title(country, fontsize=14, fontweight='bold')
    axes[i].set_xlabel('Year', fontsize=11)
    axes[i].set_ylabel('Emissions', fontsize=11)
    axes[i].set_xticks(x)
    axes[i].set_xticklabels(years, rotation=45)
    axes[i].legend()
    axes[i].grid(axis='y', alpha=0.3)
```
